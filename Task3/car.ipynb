```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"car data.csv")
df.head(10)
```

```python
df.isnull().sum().sum()
```

```python
df['Age'] = 2026 - df['Year']
```

```python
df.drop(columns=['Year'],inplace=True)
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Selling_Price'], kde=True, color='blue')
plt.title('Distribution of Selling Price')
plt.show()
```

```python
df.isnull().sum().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
x1 = df.iloc[:,:2]
x1.head()
```

```python
x2 = df.iloc[:,3:]
x2.head()
```

```python
y = df['Selling_Price']
y.head()
```

```python
x = x1.join(x2,how='outer')
x.head()
```

```python
x.shape
```

```python
df['Owner'].value_counts()
```

```python
x.info()
```

```python
x.head()
```

```python
x.info()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
LE = LabelEncoder()
```

```python
for i in x.columns:
    if x[i].dtype == 'object':
        x[i] = LE.fit_transform(x[i])
```

```python
x.info()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(x.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=9,train_size=0.8)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
LR = LinearRegression()
```

```python
LR.fit(x_train,y_train)
```

```python
y_pred = LR.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
mean_squared_error(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
cofficents = pd.DataFrame({
    'Fetures':x.columns,
    'Coff':LR.coef_
})
```
